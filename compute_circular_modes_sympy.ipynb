```python
import sympy as sp
from sympy.vector import CoordSys3D, curl, divergence
```

```python
# 1. Définition du système de coordonnées cylindriques
N = CoordSys3D('N', transformation='cylindrical', variable_names=('r', 'theta', 'z'))

# 2. Raccourcis pour les coordonnées scalaires
r, theta, z = N.r, N.theta, N.z

# Definition du champs vectoriel E
Er = sp.Function('E_r')(r, theta, z)
Et = sp.Function('E_t')(r, theta, z)
Ez = sp.Function('E_z')(r, theta, z)
E = Er * N.i + Et * N.j + Ez * N.k

# Definition du champs vectoriel H
Hr = sp.Function('H_r')(r, theta, z)
Ht = sp.Function('H_t')(r, theta, z)
Hz = sp.Function('H_z')(r, theta, z)
H = Hr * N.i + Ht * N.j + Hz * N.k
```

```python
omega, mu, epsilon, d_z = sp.symbols("omega mu epsilon delta_z", real=True)
beta = sp.symbols("beta")
eq1 = curl(E) + sp.I * omega * mu * H
eq2 = curl(H) - sp.I * omega * epsilon * E
```

```python
champs = [Er, Et, Ez, Hr, Ht, Hz]

# 2. Création automatique des règles de remplacement
regles_derivation = {}
for f in champs:
    # d/dz -> -I * beta
    regles_derivation[sp.diff(f, z)] = -sp.I * beta * d_z * f
```

```python
eq1 = eq1.subs(regles_derivation)
eq2 = eq2.subs(regles_derivation)
```

```python
eq1_r = eq1.dot(N.i)
eq1_t = eq1.dot(N.j)
eq2_r = eq2.dot(N.i)
eq2_t = eq2.dot(N.j)
```

```python
# Résolution algébrique
sol = sp.solve([eq1_r, eq1_t, eq2_r, eq2_t], (Er, Et, Hr, Ht))
```

```python
sp.simplify(sol[Er])
```

```python
kc = sp.symbols("k_c")
simplifications = {epsilon * mu * omega **2 - d_z ** 2 * beta **2: kc ** 2, d_z ** 2: 1}
for var, expr in sol.items():
    expr = sp.simplify(expr).subs(simplifications)
    sol[var] = expr
```

```python
# 5. Affichage des solutions simplifiées
for var, expr in sol.items():
    sp.pprint(sp.Eq(var, expr))
    print("\n")
```

$$E_{t}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)} = - \frac{i \left(- \mathbf{{r}_{N}} \mu \omega \frac{\partial}{\partial \mathbf{{r}_{N}}} H_{z}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)} + \beta \delta_{z} \frac{\partial}{\partial \mathbf{{theta}_{N}}} E_{z}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)}\right)}{\mathbf{{r}_{N}} k_{c}^{2}}$$


$$H_{r}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)} = - \frac{i \left(\mathbf{{r}_{N}} \beta \delta_{z} \frac{\partial}{\partial \mathbf{{r}_{N}}} H_{z}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)} - \epsilon \omega \frac{\partial}{\partial \mathbf{{theta}_{N}}} E_{z}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)}\right)}{\mathbf{{r}_{N}} k_{c}^{2}}$$


$$H_{t}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)} = - \frac{i \left(\mathbf{{r}_{N}} \epsilon \omega \frac{\partial}{\partial \mathbf{{r}_{N}}} E_{z}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)} + \beta \delta_{z} \frac{\partial}{\partial \mathbf{{theta}_{N}}} H_{z}{\left(\mathbf{{r}_{N}},\mathbf{{theta}_{N}},\mathbf{{z}_{N}} \right)}\right)}{\mathbf{{r}_{N}} k_{c}^{2}}$$


```python
m = sp.symbols('m', integer=True)
# mode TE
subs_dict = {}
subs_dict[Ez] = 0
subs_dict[Hz] = sp.besselj(m, kc * r) * sp.cos(m * theta)
sol_TE = {}
for var, expr in sol.items():
    expr = expr.subs(subs_dict).doit()
    sol_TE[var] = sp.simplify(expr)
    # 5. Affichage des solutions simplifiées
    sp.pprint(sp.Eq(var, expr))
    print("\n")
```

```python
sp.conjugate(sol_TE[Ht])
```

```python
# compute poynting vector
Sz = sp.Rational(1, 2) * (sol_TE[Er] * sp.conjugate(sol_TE[Ht]) - sol_TE[Et] * sp.conjugate(sol_TE[Hr]))
Sz = Sz.subs(sp.conjugate(theta), theta)
Sz = Sz.subs(sp.conjugate(kc), kc)
Sz = Sz.subs(sp.conjugate(sp.besselj(m - 1, r*kc)), sp.besselj(m - 1, r*kc))
Sz = Sz.subs(sp.conjugate(sp.besselj(m, r*kc)), sp.besselj(m, r*kc))
Sz = Sz.subs(sp.conjugate(sp.besselj(m + 1, r*kc)), sp.besselj(m + 1, r*kc))
Sz = Sz.subs(sp.conjugate(N.r), r)
print(sp.simplify(Sz * r))
sp.simplify(Sz * r)
```

```python
integral_theta = sp.integrate(Sz * r, (theta, 0, 2*sp.pi))
sp.simplify(integral_theta)
#print(sp.simplify(integral_theta))
```
